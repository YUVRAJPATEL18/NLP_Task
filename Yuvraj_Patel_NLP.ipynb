```python
class InputEmbedding:

    def __init__(self):
        # Sentence: "I love AI"
        # 2-dimensional embeddings

        self.X = [
            [1, 0],      # I
            [0, 1],      # love
            [1, 1]       # AI
        ]

    def display(self):
        print("Input Embeddings:")
        for row in self.X:
            print(row)


embedding = InputEmbedding()
embedding.display()
```

```python
class PositionalEncoding:

    def __init__(self):

        self.P = [
            [0.00, 1.00],
            [0.84, 0.54],
            [0.91, -0.42]
        ]

    def add_position(self, X):

        Z = []

        for i in range(len(X)):

            row = []

            for j in range(len(X[0])):

                value = X[i][j] + self.P[i][j]
                row.append(value)

            Z.append(row)

        return Z

    def display(self, Z):

        print("After Positional Encoding:")

        for row in Z:
            print(row)


position = PositionalEncoding()

Z = position.add_position(embedding.X)

position.display(Z)
```

```python
class MatrixOperations:

    def multiply(self, A, B):

        rows_A = len(A)
        cols_A = len(A[0])
        cols_B = len(B[0])

        result = []

        for i in range(rows_A):

            row = []

            for j in range(cols_B):

                total = 0

                for k in range(cols_A):

                    total = total + A[i][k] * B[k][j]

                row.append(total)

            result.append(row)

        return result


    def transpose(self, A):

        rows = len(A)
        cols = len(A[0])

        result = []

        for j in range(cols):

            row = []

            for i in range(rows):

                row.append(A[i][j])

            result.append(row)

        return result


matrix = MatrixOperations()
```

```python
class QKV:

    def __init__(self):

        # Weight matrices

        self.WQ = [
            [1, 0],
            [0, 1]
        ]

        self.WK = [
            [1, 0],
            [0, 1]
        ]

        self.WV = [
            [1, 0],
            [0, 1]
        ]


    def calculate(self, Z):

        Q = matrix.multiply(Z, self.WQ)

        K = matrix.multiply(Z, self.WK)

        V = matrix.multiply(Z, self.WV)

        return Q, K, V


qkv = QKV()

Q, K, V = qkv.calculate(Z)

print("Q:")
for row in Q:
    print(row)

print("\nK:")
for row in K:
    print(row)

print("\nV:")
for row in V:
    print(row)
```

```python
import math


class ScaledDotProduct:

    def calculate(self, Q, K):

        # K transpose

        KT = matrix.transpose(K)

        # Q × K^T

        scores = matrix.multiply(Q, KT)

        # d_k

        d_k = len(K[0])

        # sqrt(d_k)

        scale = math.sqrt(d_k)

        # Divide every element by sqrt(d_k)

        for i in range(len(scores)):

            for j in range(len(scores[0])):

                scores[i][j] = scores[i][j] / scale

        return scores


attention = ScaledDotProduct()

scores = attention.calculate(Q, K)

print("Scaled Attention Scores:")

for row in scores:
    print(row)
```

```python
class Softmax:

    def calculate(self, row):

        exp_values = []

        for x in row:

            exp_values.append(math.exp(x))

        total = 0

        for value in exp_values:

            total = total + value

        result = []

        for value in exp_values:

            result.append(value / total)

        return result


    def apply(self, matrix_scores):

        result = []

        for row in matrix_scores:

            result.append(self.calculate(row))

        return result


softmax = Softmax()

attention_weights = softmax.apply(scores)

print("Attention Weights:")

for row in attention_weights:
    print(row)
```

```python
class AttentionOutput:

    def calculate(self, attention_weights, V):

        output = matrix.multiply(
            attention_weights,
            V
        )

        return output


attention_output = AttentionOutput()

output = attention_output.calculate(
    attention_weights,
    V
)

print("Final Attention Output:")

for row in output:

    print(row)
```

```python
class Encoder:

    def __init__(self):

        self.embedding = InputEmbedding()

        self.position = PositionalEncoding()

        self.matrix = MatrixOperations()

        self.qkv = QKV()

        self.scaled_attention = ScaledDotProduct()

        self.softmax = Softmax()

        self.attention_output = AttentionOutput()


    def run(self):

        # Step 1: Input Embedding

        X = self.embedding.X

        print("STEP 1: INPUT EMBEDDING")

        for row in X:
            print(row)


        # Step 2: Positional Encoding

        Z = self.position.add_position(X)

        print("\nSTEP 2: POSITIONAL ENCODING")

        for row in Z:
            print(row)


        # Step 3: Q, K, V

        Q, K, V = self.qkv.calculate(Z)

        print("\nSTEP 3: QUERY")

        for row in Q:
            print(row)

        print("\nKEY")

        for row in K:
            print(row)

        print("\nVALUE")

        for row in V:
            print(row)


        # Step 4: Scaled Dot Product

        scores = self.scaled_attention.calculate(Q, K)

        print("\nSTEP 4: SCALED ATTENTION SCORES")

        for row in scores:
            print(row)


        # Step 5: Softmax

        weights = self.softmax.apply(scores)

        print("\nSTEP 5: SOFTMAX / ATTENTION WEIGHTS")

        for row in weights:
            print(row)


        # Step 6: Final Output

        output = self.attention_output.calculate(
            weights,
            V
        )

        print("\nSTEP 6: FINAL ENCODER OUTPUT")

        for row in output:
            print(row)


# Create Encoder

encoder = Encoder()

# Run Encoder

encoder.run()
```
